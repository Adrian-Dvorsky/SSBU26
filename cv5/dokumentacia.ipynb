## Dokumentácia k experimentom

Pred škálovaním mali jednotlivé atribúty datasetu rozdielne rozsahy hodnôt. Konkrétne atribúty mean area, area error a worst area dosahovali hodnoty rádovo v tisícoch. Väčšina ostatných atributov mala hodnoty v intervaloch rádovo 0,1 - 10

1. **StandardScaler** - Po štandardizácii majú všetky atribúty priemer okolo 0. Niektoré atribúty ale dosahujú extrémne odľahlé body dosahujúce hodnoty 10 - 12
2. **MinMaxScaler** - stlačila hodnoty od intervalu 0 - 1
3. **RobustScaler** - Podbne výsledky ako standardScalar. Odľahlým hodnoty sú najme pri atribúte area error



#### Pozorovanie pre konkrétny atribút - worst fractal dimension**

**Po škalovaní**
