## Uloha 2
PCA - na základ PCA a z grafov je viditeľné, že PC1 je najvýznamnejší, pričom vysvetľuje 24,1% celkovej variability (najväčší rozdiel mezdi control a patient). Ostatné komponenty sú menej dôležité.

K-mean - nechal som default nastavenia s 3 zhlukmi. Väčšína vzoriek zo skupiny control sa nachádza v prvom zhluku, zatiaľ čo vzorky zo skupiny patient sú v druhom zhluku. Z tohto grafu je tiež vidieť, že dáta dobre rozdeľujú chorého od zravého

PLS-DA - model dosiahol najlepšiu separáciu vskupín v rámci prvého komponentu - 23.7%. Potvrdzuje výsledky z PCA

SVM - najnižšiu chybovosť (1.6%) dosiahol model pri použití 200 premenných. Pri znížovaní počtu premenných začala chyba rásť, pričom maximum dosiahol model pri 30 premenných (10.8%).
