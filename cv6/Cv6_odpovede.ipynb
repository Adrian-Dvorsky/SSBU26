## Úloha 4

Pre porovanie výsledkov som zvolil počet replikácii = 40, oproti pôvodným 10.
Z grafu  accurancy per replication môžeme vidieť, že logistic regression dosahuje priemernú hodnotu 0.98, zatiaľ čo random forest dosahuje hodnotu 0.96. V tejto metrike dosahuje logistic regression konzistentne lepšie výsledky.

Z grafu precision per replication je podobné ako pri accurancy, logistic regression dasuhej hodnoty 0.97 a random forest 0.96

Z výsledkov confusion matrix má logistic regression výrazne menej chýb oboch typov (menej FP aj FN) a taktiež LR dosahuje vyššie hodnoty pre TP a TN.

F1-score a ROC AUC ukazujú prevahu logistic regression. Random forest zase dosahuje širšie distribúcie s nižšími vrcholmi.

**Zhrnutie**
V tomto experimente dosiahol LR lepšie výsledky porovnaní s RF. LP je presnejší, stabilnejší a má nižšiu chybovosť ako RF.
