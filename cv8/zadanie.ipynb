H0: otec je biologickým otcom dcéry
H1: otec nie je biologický otcom dcéry

Na analýzu boli využitá Česká databáza.

Pre hypotézu H0 vyšla hodnota LR(exact) = 0.108657 čo po dosadení do vzorca LR/(LR+1) výjde hodnota W = 0.098 => 9.8%. Z toho vyplíva, že pre H1 bude W=0.901 => 90.2%

Na základe výsledkov analýzy otcovstvo zamietame, pretože pre H1 je pravdepodobnosť len 9.8%, čo je veľmi nízka hodnota. Aj keď to nemôžeme potvrdiť na 100%, tak z veľkou pravdepodobnosťou muž nie je biologickým otcom dcéry.
